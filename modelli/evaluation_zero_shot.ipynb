Metriche tradizionali e BERTScore :
- BLEU/ROUGE: Contano quante parole o sequenze di parole coincidono
- BERTScore: usa il modello BERT per capire se il senso è lo stesso, anche se le parole sono diverse

```python
import evaluate
import json

bleu = evaluate.load("sacrebleu")
rouge = evaluate.load("rouge")
bertscore = evaluate.load("bertscore")

with open('risultati_zero_shot_phi4.json','r') as file:
    dati = json.load(file)

predictions = [item['generated_story']['generated_story'] for item in dati]
references = [item['reference_story'] for item in dati]

print(f"Caricate {len(predictions)} storie per la valutazione.")

results_bleu = bleu.compute(predictions=predictions, references=[[r] for r in references])
results_rouge = rouge.compute(predictions=predictions, references=references)
results_bert = bertscore.compute(predictions = predictions, references = references, lang="en")

# --- STAMPA RISULTATI PER LA TESI ---
print("\n" + "="*40)
print("   REPORT VALUTAZIONE: PHI-4 ZERO-SHOT")
print("="*40)
print(f"BLEU-4:         {results_bleu['score']:.2f}")
print(f"ROUGE-1:        {results_rouge['rouge1']:.4f}")
print(f"ROUGE-2:        {results_rouge['rouge2']:.4f}")
print(f"ROUGE-L:        {results_rouge['rougeL']:.4f}")

# Calcolo media F1 per BERTScore
media_bert = sum(results_bert['f1']) / len(results_bert['f1'])
print(f"BERTScore (F1): {media_bert:.4f}")
print("="*40)

# Opzionale: salvataggio dei risultati in un piccolo file JSON di riepilogo
final_results = {
    "bleu": results_bleu['score'],
    "rouge": results_rouge,
    "bertscore_mean": media_bert
}
with open('metrics_summary_phi4.json', 'w') as f:
    json.dump(final_results, f, indent=4)
```
